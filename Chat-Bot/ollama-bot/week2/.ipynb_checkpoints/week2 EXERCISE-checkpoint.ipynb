```python
import gradio as gr
import requests
import json
import datetime
import sqlite3
import os
import ollama 
```

```python
def initialize_database():
    if not os.path.exists('bookings.db'):
        conn = sqlite3.connect('bookings.db')
        cursor = conn.cursor()
        cursor.execute('''
        CREATE TABLE bookings (
            id INTEGER PRIMARY KEY,
            name TEXT,
            date TEXT,
            time TEXT,
            service TEXT,
            status TEXT DEFAULT 'confirmed'
        )
        ''')
        
        # Add some sample bookings
        sample_bookings = [
            ('John Doe', '2025-05-25', '10:00', 'Haircut', 'confirmed'),
            ('Jane Smith', '2025-05-26', '14:30', 'Massage', 'confirmed'),
            ('Bob Johnson', '2025-05-27', '11:15', 'Consultation', 'confirmed')
        ]
        cursor.executemany(
            'INSERT INTO bookings (name, date, time, service, status) VALUES (?, ?, ?, ?, ?)',
            sample_bookings
        )
        conn.commit()
        conn.close()
        print("Database initialized with sample bookings")
```

```python
# # Function to connect to Ollama
# def ask_ollama(prompt, model="llama3.2", ):
#     """Send a prompt to Ollama and get a response"""
    
#     try:
#         response = requests.post(
#             "http://localhost:11434/api/generate",
#             json={
#                 "model": model,
#                 "prompt": prompt,
#                 "stream": False
#             }
#         )
     
#         if response.status_code == 200:
#             return response.json()["response"]
#         else:
#             return f"Error: {response.status_code} - {response.text}"
#     except Exception as e:
#         return f"Failed to connect to Ollama: {str(e)}"



def ask_ollama(prompt, model="llama3.2"):
    """Send a prompt to Ollama and handle function call if present."""
    try:
        response = requests.post(
            "http://localhost:11434/api/generate",
            json={
                "model": model,
                "prompt": prompt,
                "stream": False,
            }
        )

        if response.status_code == 200:
            raw = response.json().get("response", "")

            # Try to parse the response as JSON if it's structured
            try:
                data = json.loads(raw)
                if "function_call" in data:
                    func = data["function_call"]
                    name = func.get("name")
                    args = func.get("arguments", {})

                    if name == "find_booking":
                        result = find_booking(**args)
                        return result
            except json.JSONDecodeError:
                pass  # Normal text response, not a structured function call

            return raw  # Just return the raw text if not a function call
        else:
            return f"Error: {response.status_code} - {response.text}"
    except Exception as e:
        return f"Failed to connect to Ollama: {str(e)}"
```

```python
find_booking_function = {
    "name": "find_booking",
    "description": (
        "Search for a user's appointment based on full name, last name, or date of appointment. "
        "Use this when the user asks something like 'Do I have an appointment?' or 'Look up John Smith on June 2nd'."
    ),
    "parameters": {
        "type": "object",
        "properties": {
            "full_name": {
                "type": "string",
                "description": "The full name of the person, e.g. 'John Smith'. Optional if last name is given.",
            },
            "last_name": {
                "type": "string",
                "description": "The last name of the person. Use this if only the surname is mentioned.",
            },
            "appointment_date": {
                "type": "string",
                "description": "Date of the appointment in YYYY-MM-DD format. Optional, used for filtering.",
            },
        },
        "required": ["full_name"],
        "additionalProperties": False
    }
}
```

```python
def find_booking(full_name=None, last_name=None, appointment_date=None):
    results = []

    for booking in bookings:
        match = True

        if full_name and full_name.lower() not in booking["name"].lower():
            match = False
        if last_name and last_name.lower() not in booking["name"].split()[-1].lower():
            match = False
        if appointment_date and appointment_date != booking["date"]:
            match = False

        if match:
            results.append(booking)

    if not results:
        return "No bookings found matching the given criteria."
    
    return results
```

### All Users Functions based on DB info and user questions ###


```python
def get_available_slots(date_str):
    # In a real application, this would check your database or calendar
    # For this example, we'll generate some fake available slots
    
    try:
        date_obj = datetime.datetime.strptime(date_str, '%Y-%m-%d')
        
        # Check if date is in the past
        if date_obj.date() < datetime.datetime.now().date():
            return []
        
        # Get booked times for this date
        conn = sqlite3.connect('bookings.db')
        cursor = conn.cursor()
        cursor.execute('SELECT time FROM bookings WHERE date = ?', (date_str,))
        booked_times = [row[0] for row in cursor.fetchall()]
        conn.close()
        
        # Generate all possible slots and remove booked ones
        all_slots = [f"{hour:02d}:{minute:02d}" for hour in range(9, 18) for minute in [0, 30]]
        available_slots = [slot for slot in all_slots if slot not in booked_times]
        
        return available_slots
    except Exception as e:
        print(f"Error getting available slots: {e}")
        return []

# Function to add a booking
def add_booking(name, date, time, service):
    try:
        conn = sqlite3.connect('bookings.db')
        cursor = conn.cursor()
        
        # Check if the slot is already booked
        cursor.execute('SELECT id FROM bookings WHERE date = ? AND time = ?', (date, time))
        if cursor.fetchone():
            conn.close()
            return False, "This time slot is already booked."
        
        # Add the booking
        cursor.execute(
            'INSERT INTO bookings (name, date, time, service, status) VALUES (?, ?, ?, ?, ?)',
            (name, date, time, service, 'confirmed')
        )
        conn.commit()
        conn.close()
        return True, "Booking confirmed!"
    except Exception as e:
        return False, f"Error adding booking: {str(e)}"

# Function to get all bookings
def get_all_bookings():
    try:
        conn = sqlite3.connect('bookings.db')
        cursor = conn.cursor()
        cursor.execute('SELECT name, date, time, service, status FROM bookings ORDER BY date, time')
        bookings = cursor.fetchall()
        conn.close()
        
        # Format the bookings as a string
        if bookings:
            result = "Current Bookings:\n\n"
            for booking in bookings:
                result += f"Name: {booking[0]}\nDate: {booking[1]}\nTime: {booking[2]}\nService: {booking[3]}\nStatus: {booking[4]}\n\n"
            return result
        else:
            return "No bookings found."
    except Exception as e:
        return f"Error getting bookings: {str(e)}"

# Function to handle the AI assistant interaction
def process_booking_request(user_message, conversation_history=""):
    # Add the user message to conversation history
    full_conversation = f"{conversation_history}\nUser: {user_message}\nAssistant: "
    
    # Create a prompt for Ollama to understand booking context
    system_prompt = """
    You are a helpful booking assistant. Help the user book appointments by:
    1. If they want to make a booking, ask for name, date, time, and service.
    2. If they want to check available slots, tell them what's available for the specified date.
    3. If they want to check existing bookings, provide the list of all bookings.
    
    If a user question is not related to bookings, politely steer the conversation back to booking-related topics.
    """
    
    # First check if the user wants to see available slots for a specific date
    date_formats = [
        "%Y-%m-%d",
        "%Y/%m/%d",
        "%d-%m-%Y", 
        "%d/%m/%Y",
        "%B %d, %Y",
        "%b %d, %Y"
    ]
    
    date_str = None
    for date_format in date_formats:
        try:
            # Look for dates in the message
            words = user_message.split()
            for i in range(len(words)):
                potential_date = " ".join(words[i:i+3])  # Check for date phrases like "May 25, 2025"
                try:
                    date_obj = datetime.datetime.strptime(potential_date, date_format)
                    date_str = date_obj.strftime("%Y-%m-%d")
                    break
                except ValueError:
                    pass
                
                # Check for single word date
                try:
                    date_obj = datetime.datetime.strptime(words[i], date_format)
                    date_str = date_obj.strftime("%Y-%m-%d")
                    break
                except ValueError:
                    pass
                    
            if date_str:
                break
        except:
            continue
    
    # Check if the user wants to see available slots
    if date_str and ("available" in user_message.lower() or "slots" in user_message.lower() or "times" in user_message.lower()):
        slots = get_available_slots(date_str)
        if slots:
            slots_text = ", ".join(slots)
            return f"Available slots for {date_str}:\n{slots_text}"
        else:
            return f"No available slots for {date_str}."
    
    # Check if the user wants to make a booking with all information
    if all(keyword in user_message.lower() for keyword in ["book", "name", "service"]) and date_str:
        # Try to extract booking information
        try:
            # Simple extraction - in a real app, you'd use NLP
            name_start = user_message.lower().find("name") + 5
            name_end = user_message.find(",", name_start)
            if name_end == -1:
                name_end = user_message.find(" at ", name_start)
            if name_end == -1:
                name_end = len(user_message)
            name = user_message[name_start:name_end].strip()
            
            time_start = user_message.lower().find("time") + 5
            if time_start < 5:  # "time" not found
                time_start = user_message.lower().find(" at ") + 4
            time_end = user_message.find(",", time_start)
            if time_end == -1:
                time_end = user_message.find(" for ", time_start)
            if time_end == -1:
                time_end = len(user_message)
            time = user_message[time_start:time_end].strip()
            
            # Format time to HH:MM
            if ":" not in time:
                if "am" in time.lower() or "pm" in time.lower():
                    # Convert "3pm" to "15:00"
                    is_pm = "pm" in time.lower()
                    time = time.lower().replace("am", "").replace("pm", "").strip()
                    hour = int(time)
                    if is_pm and hour < 12:
                        hour += 12
                    time = f"{hour:02d}:00"
                else:
                    # Assume hour only
                    time = f"{int(time):02d}:00"
            
            service_start = user_message.lower().find("service") + 8
            if service_start < 8:  # "service" not found
                service_start = user_message.lower().find(" for ") + 5
            service = user_message[service_start:].strip()
            
            # Add the booking
            success, message = add_booking(name, date_str, time, service)
            if success:
                return f"Great! I've booked {service} for {name} on {date_str} at {time}."
            else:
                return message
        except Exception as e:
            # If extraction fails, use Ollama for a more natural response
            full_prompt = f"{system_prompt}\n{full_conversation}"
            return ask_ollama(full_prompt)
    
    # Check if the user wants to see all bookings
    if "all bookings" in user_message.lower() or "show bookings" in user_message.lower() or "list bookings" in user_message.lower():
        return get_all_bookings()
    
    # For all other inquiries, use Ollama
    full_prompt = f"{system_prompt}\n{full_conversation}"
    return ask_ollama(full_prompt)

# Initialize the database
initialize_database()
```

 Creating UI for that chat 


```python
with gr.Blocks(title="Simple Booking Assistant") as app:
    gr.Markdown("# 📅 Simple Booking Assistant")
    gr.Markdown("This demo shows how to create a basic booking system with AI assistance using Ollama.")
    
    conversation_state = gr.State("")
    
    with gr.Row():
        with gr.Column():
            gr.Markdown("### Chat with the Booking Assistant")
            chatbot = gr.Chatbot(height=400)
            user_input = gr.Textbox(placeholder="Type your message here...", label="Your Message")
            clear_btn = gr.Button("Clear Conversation")
        
        with gr.Column():
            gr.Markdown("### Quick Actions")
            date_picker = gr.Textbox(label="Select Date (YYYY-MM-DD)", placeholder="2025-05-25")
            check_slots_btn = gr.Button("Check Available Slots")
            
            gr.Markdown("### Make a Booking")
            name_input = gr.Textbox(label="Name")
            date_input = gr.Textbox(label="Date (YYYY-MM-DD)")
            time_dropdown = gr.Dropdown(label="Time", choices=[])
            service_input = gr.Textbox(label="Service")
            book_btn = gr.Button("Book Appointment")
            
            view_bookings_btn = gr.Button("View All Bookings")
    
    # Update available times when date is selected
    def update_times(date):
        if date:
            return gr.Dropdown(choices=get_available_slots(date))
        return gr.Dropdown(choices=[])
    date_input.change(update_times, date_input, time_dropdown)
    
    # Handle chat messages
    def chat(user_message, history, conversation):
        
        if not user_message:
            return history, conversation, ""
        
        response = process_booking_request(user_message, conversation)
        history.append((user_message, response))
        conversation += f"\nUser: {user_message}\nAssistant: {response}"
        return history, conversation, ""
    
    # user_input.submit(chat, [user_input, chatbot, conversation_state], [chatbot, conversation_state])
    user_input.submit(chat, [user_input, chatbot, conversation_state], [chatbot, conversation_state, user_input])

    # Handle check slots button
    def check_slots(date):
        if not date:
            return [("", "Please enter a date.")]
        
        slots = get_available_slots(date)
        if slots:
            slots_text = ", ".join(slots)
            return [("", f"Available slots for {date}:\n{slots_text}")]
        else:
            return [("", f"No available slots for {date}.")]
    
    check_slots_btn.click(check_slots, date_picker, chatbot)
    
    # Handle book appointment button
    def book_appointment(name, date, time, service):
        if not all([name, date, time, service]):
            return [("", "Please fill in all fields.")]
        
        success, message = add_booking(name, date, time, service)
        if success:
            return [("", f"Booking confirmed! {name} booked for {service} on {date} at {time}.")]
        else:
            return [("", message)]
    
    book_btn.click(book_appointment, [name_input, date_input, time_dropdown, service_input], chatbot)
    
    # Handle view bookings button
    def view_bookings():
        bookings = get_all_bookings()
        return [("", bookings)]
    
    view_bookings_btn.click(view_bookings, None, chatbot)
    
    # Clear chat history
    def clear_chat():
        return [], ""
    
    clear_btn.click(clear_chat, None, [chatbot, conversation_state])

# Launch the app
if __name__ == "__main__":
    app.launch()
```
