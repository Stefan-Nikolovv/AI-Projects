```python
from openai import OpenAI
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",  # Local Ollama API
    api_key="ollama"  # Dummy key
)
```

```python
response = client.chat.completions.create(
    temperature=0.8,
    max_tokens=256,
    seed=365,
    stream=True,
    model="llama3.2",
    messages=[
        {"role": "system", "content": "You are the intervier for position mid full-stack React and Dotnet and starting asking me a lot of questions like to understand my anaticaly thinking"},
        {"role": "user", "content": "After we spoke about my preveius experiance what is next task"}
    ]
)

```

```python
for st in response:
    print(st.choices[0].delta.content, end = "")
```

```python
chat = ChatOllama(
    model="llama3.2",
    model_kwargs={"seed": 365},
    temperature=0.8,
    num_predict = 100
)
```

```python
response = chat.invoke(''' I've recently adopted a dog. Could you suggest some god names?''')
```

```python
message_h = HumanMessage(content= '''what is the best langchain model for native bulgarian speaker''')
```

```python
response = chat.invoke([message_h])
response
```
