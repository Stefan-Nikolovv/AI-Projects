```python
from openai import OpenAI
from langchain_openai.chat_models import ChatOpenAI
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",  # Local Ollama API
    api_key="ollama"  # Dummy key
)
```

```python
response = client.chat.completions.create(
    temperature=0.8,
    max_tokens=256,
    seed=365,
    stream=True,
    model="llama3.2",
    messages=[
        {"role": "system", "content": "You are a Manager at Draftkings and you are the intervier for position mid full-stack React and Dotnet and starting asking me a lot of questions like to understand my anaticaly thinking"},
        {"role": "user", "content": "After we spoke about my preveius experiance what is next task"}
    ]
)

```

```python
for st in response:
    print(st.choices[0].delta.content, end = "")
```
