```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts.chat import (SystemMessagePromptTemplate, HumanMessagePromptTemplate, ChatPromptTemplate)
from langchain_core.output_parsers import CommaSeparatedListOutputParser, StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
```

```python
chat = ChatOllama(
    model = "llama3.2:latest",
    validate_model_on_init = True,
    temperature = 0.8,
    num_predict = 256,
    seed= 365
)
```

```python
messages = [
    ("system", "You are a helpful booking company. providing information about hotels and available rooms."),
    ("human", "I am seraching a good hotel with free rooms im Bulgaria."),
]
response = chat.invoke(messages)
```

```python
response = chat.invoke(''' I've recently adopted a dog. Could you suggest some god names?''')
print(response.content)
```

# Piping a Prompt, Model and an Output Parser

```python
list_instructions = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
list_instructions
```

```python
chat_template = ChatPromptTemplate.from_messages([('human', "I've recently adeopted a {pet}. Could you suggest three {pet} names? \n" + list_instructions)])
```

```python
print(chat_template.messages[0].prompt.template )
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
chat_template_result = chat_template.invoke({'pet': 'horse'})
```

```python
chat_result = chat.invoke(chat_template_result)
```

```python
list_output_parser.invoke(chat_result)
```

```python
chat_chain = chat_template | chat | list_output_parser
```

```python
chat_chain.invoke({"pet": "dog"})
```

# Piping Chains and the RunnablePassthrought Class


```python
chat_template_tools = ChatPromptTemplate.from_template('''What are the five most importnatn tools a {job title} needs? 
Answer only by listing the tools
''')

chat_templete_strategy = ChatPromptTemplate.from_template('''Considering the tools provided, develop a strategy for effectively learing and mastering them : {tools} ''')
```

```python
string_parser = StrOutputParser()
```

```python
chain_tools = chat_template_tools | chat | string_parser | {"tools": RunnablePassthrough()}
chain_strategy = chat_templete_strategy | chat | string_parser
```

```python
print(chain_tools.invoke({"job title" : "data scientist"}))
```

```python
print(chain_strategy.invoke({'tools': '''1. Python
2. R
3. SQL
4. Tableau/Power BI
5. Excel '''}))
```

```python
chain_combined = chain_tools | chain_strategy
```

```python
print(chain_combined.invoke({"job title" : "data scientist"}) )
```

```python
chain_long = (chat_template_tools | chat | string_parser | {"tools": RunnablePassthrough()} |  chat_templete_strategy | chat | string_parser)
```

```python
chain_long.get_graph().print_ascii()
```

# RunnableParallel

```python
chat_template_books = ChatPromptTemplate.from_template(
    '''
    Suggest three of the best intermediate-level {programming language} books. 
    Answer only by listing the books.
    '''
)

chat_template_projects = ChatPromptTemplate.from_template(
    '''
    Suggest three interesting {programming language} projects suitable for intermediate-level programmers. 
    Answer only by listing the projects.
    '''
)
```

```python
chain_books = chat_template_books | chat | string_parser

chain_projects = chat_template_projects | chat | string_parser
```

```python
print(chain_books.invoke({"programming language": "Javascript"}))
```

```python
print(chain_projects.invoke({"programming language": "Javascript"}))
```

```python
chain_parallel = RunnableParallel({'books':chain_books, 'projects':chain_projects})
```

```python
chain_parallel.invoke({'programming language':'Python'})
```

```python
chain_parallel.get_graph().print_ascii()
```
