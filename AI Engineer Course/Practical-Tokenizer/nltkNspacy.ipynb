```python
from datasets import load_dataset
import pandas as pd
from nltk import ngrams
from nltk import pos_tag
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer
import spacy
import re
```

```python
dataset = load_dataset("imdb", split='train[:20]')
```

```python
df = pd.DataFrame(dataset)
```

```python
print(df)
```

```python
df['no_punkt'] = df['text'].apply(lambda x : re.sub(r'[^a-zA-Z\s]', '', x))
print(df['no_punkt'])
```

```python
df['tokens'] = df['no_punkt'].apply(lambda x : [word.lower() for word in word_tokenize(x)])
```

```python
print(df[['text', 'tokens']])
```

```python
en_stopwords = stopwords.words('english')
```

```python
df['filtered_tokens'] = df['tokens'].apply(lambda tokens: [word for word in tokens if word not in en_stopwords])
```

```python
ps = PorterStemmer()
```

```python
df['stemmed_tokens'] = df['filtered_tokens'].apply(lambda x: [ps.stem(word) for word in x])
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
df['lemmatized_tokens'] = df['filtered_tokens'].apply(lambda x: [lemmatizer.lemmatize(word) for word in x])
```

```python
unigrams = df['lemmatized_tokens'].apply(lambda x: list(ngrams(x, 1)))
```

```python
df['pos_tags'] = df['filtered_tokens'].apply(pos_tag)
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def extract_ner(text):
    doc = nlp(text)
 
    return [(ent.text, ent.label_) for ent in doc.ents]
```

```python
df['ner_entities'] = df['no_punkt'].apply(extract_ner)

```
