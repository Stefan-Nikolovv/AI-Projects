```python
df['tokenslower'] = df['text'].str.lower()
df['tokens'] = df['tokenslower'].apply(word_tokenize)
print(df['tokens'])
```

```python
en_stopwords = stopwords.words('english')
```

```python
def clean_token(tokens, stopwords_list):
    return [word for word in tokens if word.lower() not in stopwords_list]
```

```python
df["tokens_clean"] = df['tokens'].apply(lambda token: " ".join([word for word in token.split() if word not in (en_stopwords)]))
print(df["tokens_clean"])
```

```python
custom_lemmatizer = {
    'четох': 'чета',
    'важна': 'важен',
    'машинното': 'машинен'
}
```

```python
def lemmatize_tokens(tokens, lemmatizer_dict):
    lemmatized_list = []
    for word in tokens:
        lemmatized_list.append(lemmatizer_dict.get(word.lower(), word))
    return lemmatized_list
```

```python
df["lemmatize_tokens"] = df["tokens_clean"].apply(lambda token: lemmatize_tokens(token, custom_lemmatizer))
print(df["lemmatize_tokens"])
```

```python
tokens_clean = sum(df['lemmatize_tokens'], [])
unigrams = (pd.Series(nltk.ngrams(tokens_clean, 1)).value_counts())
print(unigrams)
```
